```python
error_cols = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age']
```

```python
error_cols
```

```python
for c in error_cols :
    print(c)
```

```python
for c in error_cols :
    print(df[c])
    print()
    print()
```

```python
for c in error_cols:
    q1 = df[c].quantile(0.25)
    q3 = df[c].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - (1.5 * iqr)
    upper = q3 + (1.5 * iqr)
    df[c] = df[c].map(lambda x : lower if x < lower  else upper if x > upper else x)
```

```python
plt.figure(figsize = (12, 8))
sns.boxplot(df)
plt.xticks(rotation = 90)
plt.show()
```

### **Auto Random State**

```python
score_list = []
for i in range(1000):
    Xtrain, Xtest, ytrain, ytest = train_test_split(X_ros, y_ros, test_size= 0.2, random_state=i)
    rfc1 = RandomForestClassifier()
    rfc1.fit(Xtrain, ytrain)
    sc = rfc1.score(Xtest, ytest)
    score_list.append(sc)

m = max(score_list)
rs = score_list.index(m)
print("Maximum Score : ",m)
print("Random State : ",rs)
```
